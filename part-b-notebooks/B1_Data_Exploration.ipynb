This notebook performs an exploratory data analysis of the **PURE** dataset collection,
which consists of requirements extracted from five Software Requirements Specification (SRS) documents.

### Datasets

| # | Dataset Name | Source Document |
|---|---|---|
| 1 | 0000 - gamma j | GAMMA-J Web Store SRS |
| 2 | 0000 - inventory | IUFA Inventory Management SRS |
| 3 | 2001 - space fractions | Space Fractions Learning Tool SRS |
| 4 | 2007 - get real 0.2 | Get Real CS Career Outreach SRS |
| 5 | 2010 - mashboot | MashBoot SRS |

## 1. Setup and Data Loading

```python
import json
import os
from collections import Counter

PURE_BASE = '/Users/xd/Final_Project/Final Project/raw/datasets/PURE_output'

dataset_names = [
    '0000 - gamma j',
    '0000 - inventory',
    '2001 - space fractions',
    '2007 - get real 0.2',
    '2010 - mashboot',
]

datasets = {}
for name in dataset_names:
    path = os.path.join(PURE_BASE, name, 'requirements.json')
    with open(path, 'r', encoding='utf-8') as f:
        datasets[name] = json.load(f)
    print(f'Loaded {name}: {len(datasets[name])} requirements')

print(f'Total datasets loaded: {len(datasets)}')
```

## 2. Per-Dataset Analysis

For each dataset we examine:
- Total number of requirements
- Functional vs. Non-functional ratio
- Distribution of requirement attributes

```python
for name, reqs in datasets.items():
    total = len(reqs)
    func = sum(1 for r in reqs if r.get('type') == 'functional')
    nfunc = sum(1 for r in reqs if r.get('type') == 'non-functional')
    has_notes = sum(1 for r in reqs if r.get('notes'))
    print(f'{name}:')
    print(f'  Total={total}, Functional={func}, Non-functional={nfunc}')
    print(f'  F/NF ratio: {func/nfunc:.2f}' if nfunc > 0 else '  F/NF ratio: all functional')
    print()
```

## 3. Overall Statistics

```python
all_reqs = []
for reqs in datasets.values():
    all_reqs.extend(reqs)

total = len(all_reqs)
func = sum(1 for r in all_reqs if r.get('type') == 'functional')
nfunc = sum(1 for r in all_reqs if r.get('type') == 'non-functional')
print(f'Total requirements across all datasets: {total}')
print(f'Functional: {func} ({func/total*100:.1f}%)')
print(f'Non-functional: {nfunc} ({nfunc/total*100:.1f}%)')
```

## 4. Example Records

```python
for name, reqs in datasets.items():
    print(f'\n--- {name} (first 2) ---')
    for r in reqs[:2]:
        print(f"  {r['requirement_id']}: {r['title']}")
        print(f"    Type: {r['type']}")
        desc = r['description'][:120]
        print(f"    Desc: {desc}...")
        ac = len(r.get('acceptance_criteria', []))
        print(f"    Acceptance criteria: {ac}")
        print(f"    Source: {r.get('source_location', {}).get('document', 'unknown')}")
```

## 5. Structural Analysis

### 5.1 Description Length Distribution

```python
import statistics

for name, reqs in datasets.items():
    lengths = [len(r['description']) for r in reqs]
    print(f'{name}:')
    print(f'  Mean={statistics.mean(lengths):.0f}, 
        f'Median={statistics.median(lengths):.0f}')
    print(f'  Min={min(lengths)}, Max={max(lengths)}')
```

### 5.2 Acceptance Criteria Count

```python
for name, reqs in datasets.items():
    ac_counts = [len(r.get('acceptance_criteria', [])) for r in reqs]
    print(f'{name}: mean={statistics.mean(ac_counts):.1f}, max={max(ac_counts)}')
```

### 5.3 Source Location Presence

```python
for name, reqs in datasets.items():
    with_src = sum(1 for r in reqs if r.get('source_location') and r['source_location'].get('verse'))
    print(f'{name}: {with_src}/{len(reqs)} ({with_src/len(reqs)*100:.0f}%) have source verses')
```

## 6. Data Characteristics

The PURE dataset requirements follow a consistent schema:

| Field | Type | Description |
|---|---|---|
| requirement_id | string | Sequential ID (REQ-001, REQ-002) |
| title | string | Short descriptive title |
| description | string | Full requirement statement |
| type | string | 'functional' or 'non-functional' |
| acceptance_criteria | array | Testable acceptance criteria |
| source_location | object | Document name + verbatim quotes |

**Key observations:**
- All requirements have IDs, titles, descriptions, and types
- Source location includes verbatim SRS quotes (verse) for traceability
- No page numbers in the original PURE format (added by our extractor)
- Type distribution varies by domain (more functional in business apps)
- Description lengths range from ~50 to ~500 characters

## 7. Sensitive Features & Fairness

### 7.1 Direct Sensitive Attributes

The dataset does **not** contain sensitive attributes such as:
- Demographics (age, gender, race, ethnicity)
- Socioeconomic indicators
- Geographic identifiers tied to persons

This is expected since SRS documents describe system behavior, not user demographics.

### 7.2 Potential Bias Considerations

**Domain bias:** All datasets are from academic/university project SRS documents.

**Linguistic bias:** All documents are in English, using IEEE-style SRS conventions.

**Target audience bias:** Some requirements reference specific user groups (e.g., teens, CS students).

**Mitigation recommendations:**
1. Include SRS documents from diverse domains (healthcare, finance, government)
2. Include documents in multiple languages
3. Test extraction on SRS documents from non-academic contexts
4. Ensure the evaluation set represents varied SRS writing styles
